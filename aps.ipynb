```python
import numpy as np
np.random.seed(42)
p = 0.30
sample_size = 10
samples = np.random.binomial(n=1, p=p, size=sample_size)
print(samples)
```

```python
number_of_clicks=np.sum(samples==1)
number_of_non_clicks=np.sum(samples==0)
print(number_of_clicks)
print(number_of_non_clicks)
```

```python
empirical_p1 = number_of_clicks / sample_size
empirical_p0 = number_of_non_clicks / sample_size
print(empirical_p1)
print(empirical_p0)
```

```python
import matplotlib.pyplot as plt

# Theoretical PMF
theoretical = [1 - p, p]

# Empirical PMF
empirical = [number_of_non_clicks / sample_size, number_of_clicks / sample_size]

x = [0, 1]

plt.bar(x, theoretical, width=-0.3, align='edge', label='Theoretical')
plt.bar(x, empirical, width=0.3, align='edge', label='Empirical')

plt.xticks([0, 1], ['No Click (0)', 'Click (1)'])
plt.ylabel('Probability')
plt.title('Empirical vs Theoretical PMF')
plt.legend()
plt.show()
```

```python
empirical_var = np.var(samples)
print("Empirical Variance:", empirical_var)
```

```python
empirical_mean = np.mean(samples)
print("Empirical Expectation:", empirical_mean)
```
